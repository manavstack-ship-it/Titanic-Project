```python
import pandas as pd
import numpy as np
import seaborn as sea
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('cleaned_titanic.csv')
```

```python
df.head(15)
```

Univeriate **Analysis**

```python
from matplotlib.axis import XTick
ax = sea.countplot(x='sex', data=df, width=0.4 )
ax.set_xticks([0, 1])
ax.set_xticklabels(['Girl', 'Boy'])
sea.set_style('whitegrid')
```

```python
#pclass
sea.countplot(x='pclass', data=df, width=0.4 )
```

```python
#Survived
a=sea.countplot(x='survived', data=df, width=0.4)
a.set_xticks([0, 1])
a.set_xticklabels(['Died', 'Survived'])
```

```python
#adult_male
sea.countplot(x='adult_male', data=df, width=0.4 )
```
